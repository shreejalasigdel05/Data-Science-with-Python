```python
# heatmap shows a correlation matrix, the relationship between numerical variables,while headmap visualizes
#that matrix using colors
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tips=sns.load_dataset("tips")
df=tips.head()
df
```

```python
corr=tips.corr(numeric_only=True) #printing only numeric values from the table by correlation formula
corr
```

```python
tips=sns.load_dataset("tips")
corr=tips.corr(numeric_only=True)
plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True,cmap='coolwarm',fmt='.2f',linewidth=1, alpha=0.999)
plt.title("Tips correlation matrix")
plt.show()
```

```python
df=pd.DataFrame({
    "Age":[22,25,30,35,40],
    "Salary":[25000,30000,45000,55000,70000],
    "Experience":[1,3,5,8,12]
})
df
```

```python
corr=df.corr()
corr
```

```python
corr=df.corr()
plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True,fmt='.2f',linewidth=1, alpha=0.999)
plt.title("Employee Info")
plt.show()
```
