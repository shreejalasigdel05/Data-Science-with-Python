```python
#grouping and aggragation
import pandas as pd
```

```python
df=pd.DataFrame({
    "Department":['IT','HR','IT','Finance','HR','Finance'],
    "Employee":['Rakhi','Sikha','Anisha','Priya','Jhyoti','Ramila'],
    "Salary":[5000,6000,7000,8000,5500,9000],
    "Age":[25,30,28,35,32,40]
})
```

```python
sal_sum=df.groupby("Department")["Salary"].sum()
```

```python
sal_sum
```

```python
print(type(sal_sum))
```

```python
dept_count=sal_sum=df.groupby("Department")["Salary"].count()
```

```python
dept_count
```

```python
sal_mean=df.groupby("Department")["Salary"].mean()
```

```python
sal_mean
```

```python
sal_min=df.groupby("Department")["Salary"].min()
```

```python
sal_min
```

```python
sal_max=df.groupby("Department")["Salary"].max()
```

```python
sal_max
```

```python
result=df.groupby("Department").agg({
    "Salary":['sum','mean','max'],
    "Age":['min','max']
})
result
```

```python
#total sal of IT dept
total_IT_dept_sal=df[df['Department']=='IT']['Salary'].sum()
total_IT_dept_sal
```

```python
avg_IT_dept_sal = df[df['Department'] == 'IT']['Salary'].mean()
avg_IT_dept_sal
```

```python
#calculate the difference of salary between IT dept and finance dept
it=df[df['Department']=='IT']['Salary'].sum()
fin=df[df['Department']=='Finance']['Salary'].sum()
diff=it-fin
diff
```

```python

Gender={
    "Rakhi":"Female",
    "Sikha":"Female",
    "Anisha":"Female",
    "Priya":"Female",
    "Jhyoti":"Female",
    "Ramila":"Female"
}
```

```python
df['Gender']=df['Employee'].map(Gender)
```

```python
df
```

```python
#sum on basis of dept and gender
res=df.groupby(["Department","Gender"])["Salary"].sum()
res
```

```python
result=(
    df.groupby("Department").agg(
        avg_sal=("Salary","mean"),
        emp_count=("Employee","count")
    )
    .reset_index()
)
result
```
