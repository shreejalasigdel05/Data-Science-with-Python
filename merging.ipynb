```python
#merging
import pandas as pd
employees=pd.DataFrame({
    "EmpID":[1,2,3],
    "Name":["Rakesh","Sikha","Anil"]
})
salaries=pd.DataFrame({
    "EmpID":[1,2,4],
    "Salary":[5000,7000,6000]
})
```

```python
mdf=pd.merge(employees,salaries,on="EmpID")
mdf
```

```python
ldf=pd.merge(employees,salaries,on="EmpID",how="left")
ldf
```

```python
rdf=pd.merge(employees,salaries,on="EmpID",how="right")
rdf
```

```python
df1=pd.DataFrame({
    "Name":["Rakesh","Sikha"],
    "Age":[25,30]
})
df2=pd.DataFrame({
    "Name":["Anil","Priya"],
    "Age":[22,28]
})
```

```python
cdf=pd.concat([df1,df2],ignore_index=True)
cdf
```

```python
employees=pd.DataFrame({
    "EmpID":[1,2,3],
    "Name":["Rakesh","Sikha","Anil"]
})
salaries=pd.DataFrame({
    "EmpID":[1,2,4],
    "Salary":[5000,7000,6000]
})
```

```python
outer_merge=pd.merge(employees,salaries,on="EmpID",how="outer")
outer_merge
```

```python
left=pd.DataFrame({
    "Name":["Rakesh","Sikha","Anil"]
},index=[1,2,3])
right=pd.DataFrame({
    "Salary":[5000,7000,6000]
},index=[1,2,4])
```

```python
df=left.join(right)
df
```

```python
inner=left.join(right,how="inner") #only common
inner
```

```python
left.join(right,how="right")
```

```python
left.join(right,how="outer") #all
```

```python
order=pd.DataFrame({
    "order_id":[1,2,3,4],
    "customer_id":[101,102,103,104],
    "amount":[250,150,300,200]
})
customers=pd.DataFrame({
    "customer_id":[101,102,105],
    "name":["Alice","Bob","Eve"]
})
```

```python
cdf=pd.concat([order,customers],ignore_index=True)
cdf
```

```python
mdf=pd.merge(order,customers,on="customer_id")
mdf
```
