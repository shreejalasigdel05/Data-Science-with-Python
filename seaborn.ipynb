```python
import matplotlib.pyplot as plt
import seaborn as sns
df=sns.load_dataset('tips')
df
```

```python
plt.scatter(df['total_bill'],df['tip'])
plt.grid(True)
plt.show()
```

```python
tips=sns.load_dataset('tips')
sns.scatterplot(data=tips,x='sex',y='tip',hue='sex')
plt.grid(True)
plt.show()
```

```python
tips=sns.load_dataset('tips')
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='smoker')
plt.grid(True)
plt.show()
```

```python
sns.lineplot(data=df,x='size',y='tip')
plt.title('size vs tip')
plt.show()
```

```python
sns.scatterplot(data=df,x='day',y='tip',hue='day')
plt.title('day vs tip')
plt.show()
```
