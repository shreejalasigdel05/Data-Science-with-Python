```python
class Student:
    name='ritu'
    age=20
    fee=10000
    def sdetail(self):
        print(self.name)
        print(self.age)

    def fee_detail(self):
        val=self.fee *0.1
        total=self.fee +val
        print(f'10% of fee is {val}')
        print(f'total fee is {total}')
    def fee_reg(self,charge):
        total=self.fee+charge
        print(f'Registration fee is {total}')
s=Student()
s.sdetail()
s.fee_detail()
s.fee_reg(4500)
```

```python
class Employee:
    name="hari"
    dept='IT'
    salary=20000
    def display_detail(self):
        print(self.name)
        print(self.dept)
    def display_salary(self):
        print(self.salary)
    def display_bonus(self,bonus):
        print(f'Bonus is {bonus}')
        total_sal=self.salary+bonus
        print(f'total salary is {total_sal}')
e=Employee()
e.display_detail()
e.display_salary()
e.display_bonus(5000)
```
