```python
#data transformation
import pandas as pd
```

```python
df=pd.DataFrame({
    'Name':['Kali','Sikha','Anisha','Priya','Anya','Anya'],
    'Age':[25,30,22,28,35,5],
    'Salary':[5000,7000,4500,6500,8000,2000],
    'City':['Rjb','Brt','Ktm','Pkr','Ktm','Pkr']
})
```

```python
df
```

```python
df['Bonus']=df['Salary']*0.10 #add bonus in table
```

```python
df
```

```python
df['Name']=df['Name'].str.upper() #name in capital letter
```

```python
df
```

```python
df['Name']=df['Name'].str.lower() #lower case letter
```

```python
df
```

```python
df['Name']=df['Name'].str.title() #first letter capital all small
```

```python
df
```

```python
df['City']=df['City'].replace('Ktm','Ktm capital') #replace the cell with different value
```

```python
df
```

```python
gender={
    'Kali':'Female',
    'Sikha':'Female',
    'Anisha':'Female',
    'Priya':'Female',
    'Anya':'Female'
}
```

```python
df['Gender']=df['Name'].map(gender)
```

```python
df
```

```python
df['City']=df['City'].replace('Ktm capital','Ktm')
df
```

```python
df['Name'].unique()
```

```python
df['Age_Months']=df['Age']*12
df
```

```python
sorted_df=df.sort_values(by='Age',inplace=True) #sorting (by default its always ascending)
sorted_df
```

```python
sorted_df=df.sort_values(by='Age',ascending=False) #sorting in descending
sorted_df
```

```python
df.sort_values(by=['Age','Salary'],ascending=[False,True],inplace=True)
```

```python
df
```

```python
df.sort_values(by=['Salary','Age'],ascending=[True,False],inplace=True) #inplace is for placing the changes data in the dataframe
```

```python
df
```

```python
sorted_df=df.sort_values(by='City') 
sorted_df
```

```python
df=df.sort_values(by='Age').reset_index(drop=True)
df
```

```python
df=df[df['Age']>25]
```

```python
df
```

```python
df=df[(df['Age']>25) & (df['Salary']>5000)]
df
```

```python
#display all record from the df where age is greater than 22 and city is ktm
```

```python
df=df[(df['Age']>22) & (df['City']=='Ktm')]
df
```
