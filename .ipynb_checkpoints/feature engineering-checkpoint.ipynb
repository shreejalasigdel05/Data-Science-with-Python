```python
# feature selection
# process of selecting the useful features from the group of features
# -reduce unnecessary features
# -reduce dimension
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.DataFrame({
    "Age":[25,np.nan,30],
    "Salary":[50000,60000,70000]
})
df
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
df
```

```python
df=pd.DataFrame({
    "Age":[20,np.nan,30,28,35,np.nan],
    "Salary":[50000,60000,70000,55000,900000,65000000],
    "Gender":['male','female','male','female','female','male'],
    "Department":['HR','IT','Finance','IT','HR','Marketing'],
    "City": ['kathmandu','pokhara','kathmandu','pokhara','lalitpur','birgunj']
})
df
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
df
```

```python
#create new feature
df['Age_Group']=pd.cut(
    df['Age'],
    bins=[0,18,30,50,100],
    labels=['young','adult','middle_age','senior']
)
df[['Age','Age_Group']]
```

```python
df['Salary_Level']=pd.cut(
    df['Salary'],
    bins=[0,60000,1000000,float('inf')],
    labels=['low','medium','high']
)
df[['Salary','Salary_Level']]
```

```python
df
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder(drop='first',sparse_output=False)
encoded=encoder.fit_transform(df[['Gender','Department','City']])
encoded
```

```python
df_encoded=pd.get_dummies(
    df,
    columns=['Gender','Department','City','Age_Group','Salary_Level'],
    dtype=int
)
df_encoded
```

```python
encoded_df=pd.DataFrame(encoded, columns=encoder.get_feature_names_out(['Gender','Department','City']))
encoded_df
```
