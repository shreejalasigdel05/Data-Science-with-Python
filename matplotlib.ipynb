```python
import matplotlib.pyplot as plt
```

```python
revenue=[45000,47000,55000,50000,46000]
year=[2015,2016,2017,2018,2019]
plt.plot(year,revenue) #for plotting graph
```

```python
expenses=[20000,25000,35000,40000,42000]
revenue=[45000,47000,55000,50000,46000]
year=[2015,2016,2017,2018,2019]
plt.plot(year,revenue) #1st- x axis and 2nd- y axis
plt.plot(year,expenses)
plt.style.use('default')#design 
plt.xticks(year) #showing exact value as data in graph
plt.show()
```

```python
expenses=[20000,25000,35000,40000,42000]
revenue=[45000,47000,55000,50000,46000]
year=[2015,2016,2017,2018,2019]
plt.plot(year,revenue, color='red') 
plt.plot(year,expenses, color='yellow')
plt.title('Revenue & Expense over year') #giving title
plt.xlabel('Year') #give name to x axis
plt.ylabel('Revenue & Expenses') #give name to y axis
plt.style.use('default') 
plt.xticks(year) 
plt.show()
```

```python
expenses=[20000,25000,35000,40000,42000]
revenue=[45000,47000,55000,50000,46000]
year=[2015,2016,2017,2018,2019]
plt.plot(year,revenue, color='red', linestyle='dashed', linewidth=3, marker='o',label='revenue') 
plt.plot(year,expenses, color='green',linestyle='dotted', linewidth=3, marker='D',markersize='5',label='expenses')
plt.title('Revenue & Expense over year')
plt.xlabel('Year')
plt.ylabel('Revenue & Expenses') 
plt.style.use('default') 
plt.xticks(year) 
plt.legend(loc='best') #denotes which line do what
plt.grid(True,linestyle='--')
plt.show()
```
