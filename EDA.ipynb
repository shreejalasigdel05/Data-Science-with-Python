```python
# EDA- exploratory data analysis and pre-processing
# it is the process of understanding dataset before applying machine learning or statistical model
#Numerical data=continous,
# Discrete data=not continous
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.DataFrame({
    "Height":[170.5,168.3,180.9,175.1,169.8],
    "Weight":[65.2,70.1,82.5,80.40,85.0],
    "Temperature":[36.8,37.2,36.5,40.0,42.0]
}) 
df.dtypes 
```

```python
plt.scatter(df["Height"],df["Weight"])
plt.title('Height vs weight')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.grid(True)
plt.show()
```

```python
df=pd.DataFrame({
    "Department":[
        "IT","HR","IT","Sales","HR","IT","Sales","HR","IT","Sales"
    ]
})
```

```python
sns.countplot(data=df,x="Department")
plt.title('Employee count by department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.show()
```

```python
df=pd.DataFrame({
    "Department":["IT","HR","Sales","Finance"],
    "Employee":[45,20,35,25]
})
df
```

```python
sns.barplot(data=df,x="Department",y="Employee")
plt.title('Bar Chart employees count by department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.show()
```

```python
# histogram shows the distribution of numerical data by grouping them into groups called bins
df=pd.DataFrame({
    "Marks":[45,50,55,75,80,85,105,110,115]
})
```

```python
plt.hist(df["Marks"],bins=6)
plt.title('Distribution of student marks')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()
```

```python
#categorical data types
#1. nominal data-not order, eg: male/female, black/white (pie chart suitable)
#2. audinal data- have order, eg:large,very large
```

```python
#nominal data
departments=["IT","HR","Sales","Finance"]
employees=[45,20,35,25]
plt.figure(figsize=(6,6))
plt.pie(
    employees,
    labels=departments,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Employee Distribution by Department")
plt.axis("equal")
plt.show()
```
