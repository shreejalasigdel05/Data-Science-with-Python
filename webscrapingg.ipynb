```python
import requests
from bs4 import BeautifulSoup
```

```python
url='https://techaxis.com.np/'
response=requests.get(url)
response
```

```python
soup=BeautifulSoup(response.text,'html.parser')
print(soup.title.text)
```

```python
print(soup.body.text)
```

```python
h1=soup.find('h1')
if h1:
    print(f'h1 tag found:{h1.text}')
else:
    print('No h1 tag found')
```

```python
h2=soup.find('h2')
if h2:
    print(f'h2 tag found:{h2.text}')
else:
    print('No h2 tag found')
```

```python
h3=soup.find('h3')
if h3:
    print(f'h3 tag found:{h3.text}')
else:
    print('No h3 tag found')
```

```python
h2_tags=soup("h2")
for i, tag in enumerate(h2_tags,start=1):
    print(f"{i}.{tag.get_text(strip=True)}")
print(len(h2_tags))
```

```python
h2_tags=soup("h3")
for i, tag in enumerate(h2_tags,start=1):
    print(f"{i}.{tag.get_text(strip=True)}")
print(len(h2_tags))
```

```python
url="https://techaxis.com.np/"
res=requests.get(url)
soup=BeautifulSoup(res.text,"html.parser")
tags=soup.find_all(['h1','h2','h3'])
for tag in tags:
    print(tag.name,":",tag.get_text(strip=True))
```

```python
url="https://techaxis.com.np/"
res=requests.get(url)
soup=BeautifulSoup(res.text,"html.parser")
tags=soup.find_all(['p'])
for tag in tags:
    print(tag.name,":",tag.get_text(strip=True))
```
