```python
# Grid SearchCV is a scikit-learn library used to setup grids,
# settings, or hyperparameters, for a machine learning model
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
df=pd.read_csv("customer_purchase.csv")
df.head()
```

```python
plt.figure(figsize=(4,3))
sns.heatmap(df.corr(),annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
X=df[["Age","Salary","Experience"]]
y=df["Purchased"]
X_train,X_test,y_train,y_test=train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)
```

```python
rf=RandomForestClassifier()
```

```python
# hyper parameter grid
param_grid={
    "n_estimators":[10,50,100],
    "max_depth":[None,3,5],
    "min_samples_split":[2,4,],
    "min_samples_leaf":[1,2],
    "criterion":['gini','entropy']
}
```

```python
grid_search=GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=3,
    scoring= "accuracy",
    n_jobs=-1
)
```

```python
grid_search.fit(X_train,y_train)
```

```python
rf.fit(X_train, y_train)
```

```python
y_pred=rf.predict(X_test)
y_pred
```

```python
best_model=grid_search.best_estimator_
```

```python
best_model
```

```python
best_model.predict(X_test)
```

```python
y_test_array=np.array(y_test)
y_test_array
```

```python
print(grid_search.best_params_)
```

```python
print(grid_search.best_score_)
```

```python
accuracy=accuracy_score(y_test,y_pred)
print(accuracy)
```

```python
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
print(classification_report(y_test,y_pred))
```

```python
new_customer=pd.DataFrame({
    "Age":[38],
    "Salary":[58000],
    "Experience":[9]
})
prediction=best_model.predict(new_customer)
prediction
```

```python
from sklearn.tree import plot_tree
tree=best_model.estimators_[0]
plt.figure(figsize=(15,10))
plot_tree(
    tree,
    feature_names=X.columns,
    class_names=['No purches','Purchase'],
    filled=True,
    rounded=True,
    impurity=True,
    precision=2,
    fontsize=10
)
plt.title("Decision Tree from Best Random Forest Model")
plt.show()
```
