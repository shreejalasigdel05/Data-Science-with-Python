```python
#encoding 
#encoding catergorical data: one hot encoding and label encoding::
#categorical data is data that contains categories (text values) instead of numbers
#machine learning algorithms cannot work directly with text, so we convert categorical values into numbers using encoding
# 1. One-hot Encoding(for normal data)
# 2. Ordinal Encoding(for ordinal data)
# !pip install scikit-learn
```

```python
import pandas as pd
df=pd.DataFrame({
    "Color":['red','blue','green','blue','red']
})
df
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder=OneHotEncoder(sparse_output=False)
encoded=encoder.fit_transform(df[['Color']])
encoded
```

```python
encoded_df=pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(["Color"])
)
encoded_df
```

```python
encoder=OneHotEncoder(drop="first",sparse_output=False)
encoded=encoder.fit_transform(df[["Color"]])
encoded
```

```python
from sklearn.preprocessing import OrdinalEncoder
df=pd.DataFrame({
    "Size":['small','medium','large','medium','small']
})
df
```

```python
encoder=OrdinalEncoder(
    categories=[['small','medium','large']]
)
df['size']=encoder.fit_transform(df[["Size"]])
df
```

```python
df=pd.DataFrame({
    "result":['pass','fail','pass','fail','pass']
})
df
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
df["Result"]=encoder.fit_transform(df['result'])
df
```

```python
decoded=encoder.inverse_transform(df["Result"])
decoded
```

```python
df=pd.DataFrame({
    "Gender":['male','female','female','male'],
    "Size":['small','medium','large','medium'],
    "Result":['pass','fail','pass','dropout']
})
```

```python
ohe=OneHotEncoder(drop='first',sparse_output=False)
gender_encoded=ohe.fit_transform(df[['Gender']])
gender_encoded
```

```python
gender_df=pd.DataFrame(
    gender_encoded,
    columns=ohe.get_feature_names_out(["Gender"])
)
```

```python
oe=OrdinalEncoder(categories=[['small','medium','large']])
df['size']=oe.fit_transform(df[['Size']])
df
```

```python
le=LabelEncoder()
df['result']=le.fit_transform(df["Result"])
df
```

```python
df=pd.concat([df.drop("Gender",axis=1),gender_df],axis=1)
df
```
