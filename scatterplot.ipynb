```python
import matplotlib.pyplot as plt
import pandas as pd
study_hours=[1,2,3,4,5,6,7]
exam_scores=[55,60,65,72,78,85,90]
plt.figure(figsize=(6,4))
plt.scatter(study_hours,exam_scores)
plt.title('exam vs marks')
plt.xlabel('study hours')
plt.ylabel('exam scores')
plt.grid(True)
plt.show()
```

```python
#set this study hoours and marks in the data frame
```

```python
df=pd.DataFrame({
    'StudyHours':study_hours,
    'Scores':exam_scores
})
df
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['StudyHours'],df['Scores'])
plt.title('exam vs marks')
plt.xlabel('study hours')
plt.ylabel('exam scores')
plt.grid(True)
plt.show()
```
