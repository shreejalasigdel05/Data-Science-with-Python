```python
#IQR is used when data distribution is skewed or data is not 
#normal distribution
#BOX plot and IQR is used
#Interquartile Range(IQR) is measure of the spread (dispersion) of data.
```

```python
#Q1 (First Quartile)=25th percentile (25% of the data lies below this value
#Q2 (Median)= 50th percentile
#Q3 (Third Quartile)=75th percentile (75% of the data lies below this value)
# formula: IQR=Q3-Q1
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data=[4,6,8,9,10,12,14,18,20]
Q1=np.percentile(data,25)
Q2=np.percentile(data,50)
Q3=np.percentile(data,75)
```

```python
print(Q1)
```

```python
IQR=Q3-Q1
IQR
```

```python
df=pd.DataFrame({
    "Student":["A","B","C","D","E","F","G","H","I","J"],
    "Marks":[5,50,55,60,62,65,68,70,72,150]
})
df
```

```python
plt.boxplot(df["Marks"])
```

```python
plt.scatter(df["Student"],df["Marks"])
plt.title("Student vs Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
Q1=df["Marks"].quantile(0.25)
Q2=df["Marks"].quantile(0.50)
Q3=df["Marks"].quantile(0.75)
IQR=Q3-Q1
```

```python
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
lower_bound,upper_bound
```

```python
outliers=df[(df["Marks"]<lower_bound) |(df["Marks"]>upper_bound)]
outliers
```

```python
clean_df=df[(df["Marks"]>=lower_bound) & (df["Marks"]<=upper_bound) ]
clean_df
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(clean_df["Marks"],vert=True)
plt.title("After removing outliers")
plt.ylabel("Marks")
plt.show()
```
