Web scraping
Introduction to web scraping and its ethics::
Web scraping is the process of automatically extracting data from website using software or scripts instead of 
manually copying information

Sends a request to a website
Downloads the webpage's html content
Extracts the required information
Stores the data in a structured format such as CSV,Excel, or a database.


Example
Suppose you want to college product names and prices from a e-commerce website.
Instead of manually copying hundreds of products , a scraper can automatically extract
Product     Price
Laptop A    $500
Laptop B    $600
Laptop C    $700

Common python libraries for web scraping
request-downloads webpage content
beautifulshoup=parses html and extracts data
Scrapy- powerful web scraping framework
selenium-automates browsers and handles javascript-heavy websites.
pandas- stores and analyzes scraped data.

Ethics of web scraping::
Just because data is visible on a website does not always mean it should be scraped
without consideration.
1. Respect website terms of service
many websites specify whether scraping is allowed in their temrs of service.
read the website's policies.
follow any restrictions they provide.


2. check robots.txt
many websites publish a robots.txt file indicating which areas automated systems should avoid.
(robots.txt is a plain text file placed at the root of a website(for example,
https://rakesh.com/robots/txt) that tells web crawlers and search enging bots which parts
of the site they are allowed or not allowed to access.)
Example:

User-agent: *
Disallow: /admin/
Disallow: /private/
Allow: /

Sitemap: https://example.com/sitemap.xml
---------------------
https://rakesh.com/robots.txt
A robots.txt file may contain:
User-agent: *
Disallow: /private/

3. Common Directives::
User-agent- specifies which crawler the rules apply to(*mean all crawlers).
Disallow- tells crawlers not to visit a path
Allow- Explicitly permits access to a path.
Sitemap- points crawlers to the site's sitmap.

4. Respect privacy:: Donot collect: Personal information, passwords, private user data
sensitive information
unless you have proper authorization and a legitimate reason

5. Use Public APIs when available::
Many websites provide APIs specifically for accessing data.
Benefits: More reliable, faster, usually allowed by the provider, structured data format(JSON)

6. Follow copyright and data usage rules::
Even if you can scrape content, you may not have the right to republish or redistribute it.
Always check: Copyright policies, licensing terms, data usage restrictions
